# 🧪 Exercise 1: Traditional vs. Modern NLP — Comparative Analysis

## 1️⃣ Comparison Table

| **Aspect** | **Traditional NLP** | **Modern NLP (Transformers / LLMs)** |
|-------------|--------------------|--------------------------------------|
| **Feature Engineering** | Manual: hand-crafted features (n-grams, TF-IDF, POS tags). | Automatic: features learned directly from data. |
| **Word Representations** | Static embeddings (Word2Vec, GloVe). | Contextual embeddings (BERT, GPT). |
| **Model Architectures** | Shallow models (Naïve Bayes, SVM, CRF). | Deep models (Transformers, 12-100 + layers). |
| **Training Methodology** | Task-specific training on labeled data. | Pre-training + Fine-tuning paradigm. |
| **Examples** | Naïve Bayes, CRF, LSTM. | BERT, GPT-x, T5, RoBERTa. |
| **Advantages** | Simple, interpretable, works with small data. | High accuracy, transferable, minimal manual work. |
| **Disadvantages** | Needs manual features, brittle, limited context. | Compute-intensive, opaque, data hungry. |

---

## 2️⃣ Scalability & Efficiency Impact

- **Reusability**: Pre-trained backbones can be reused across tasks.  
- **Data Efficiency**: Less labeled data needed.  
- **Computation**: Heavy training once; light fine-tuning later.  
- **Maintenance**: One model → many downstream apps.

# 🌟 Exercise 2: LLM Architecture & Application Scenarios

| **Model** | **Architecture Type** | **Pre-training Objective** | **Excels At** | **Why It Fits** |
|------------|----------------------|-----------------------------|----------------|-----------------|
| **BERT** | Encoder-only (bidirectional) | Masked Language Modeling (MLM) | Classification, NER, QA | Reads full context from both directions. |
| **GPT** | Decoder-only (unidirectional) | Causal Language Modeling (next-token prediction) | Text generation, chatbots, code | Naturally produces fluent, sequential text. |
| **T5** | Encoder-Decoder | Span corruption (Text-to-Text) | Translation, summarization | Converts any task into text-to-text form. |

# 🌟 Exercise 3: Benefits & Ethics of Pre-Training

### ✅ Five Key Benefits
1. **Generalization** – understands many topics.  
2. **Less Labeled Data** – learns from unlabeled text.  
3. **Faster Fine-tuning** – needs fewer epochs.  
4. **Transfer Learning** – one model → many tasks.  
5. **Robustness** – handles rare words & noise.

### ⚠️ Ethical Concerns
- **Bias** – reflects social & gender stereotypes.  
- **Misinformation** – can generate false facts.  
- **Misuse** – spam, scams, fake content.  
- **Privacy/IP** – sensitive or copyrighted data.

### 🧩 Mitigation Strategies
- Careful **data curation & documentation**.  
- **Bias audits** and fairness metrics.  
- **Human-in-the-loop** moderation.  
- **Policy compliance** + content filtering.

# 🌟 Exercise 4: Transformer Architecture Deep Dive

## 1️⃣ Self-Attention & Multi-Head Attention

- **Self-Attention** computes weights between every token pair → each token “looks” at others.  
- **Multi-Head Attention** = several attentions in parallel; each head learns different relations (syntax, semantics, etc.).

**Example sentence:**  
> “Ada mailed the package to Ben on Monday.”  
- Head 1: Ada ↔ mailed  
- Head 2: mailed ↔ package  
- Head 3: to ↔ Ben  
- Head 4: on ↔ Monday  

---

## 2️⃣ Pre-Training Objectives

| **Objective** | **Description** | **Use Case** |
|----------------|-----------------|---------------|
| **MLM** | Mask 15% of tokens → predict missing words (bidirectional). | Understanding tasks (BERT). |
| **CLM** | Predict next token from left context only. | Text generation (GPT). |

- **NSP (Next Sentence Prediction)** was used in early BERT to link sentences; later removed (RoBERTa) because it did not help much.

---

## 3️⃣ Model Selection by Application

| **Task** | **Best Model Type** | **Reason** |
|-----------|---------------------|-------------|
| Sentiment analysis | Encoder-only (BERT) | Strong context understanding. |
| Chatbot | Decoder-only (GPT) | Generates coherent responses. |
| Translation | Encoder-Decoder (T5) | Maps input → output sequence. |

---

## 4️⃣ Positional Encoding

- Transformers lack word order awareness → **positional encoding** adds sequence order info.  
- Without it:  
  - “dog bites man” vs. “man bites dog” would look identical.

# 🌟 Exercise 5: BERT Variations — Choose Your Detective

## 1️⃣ Scenario Matching

| **Scenario** | **Best Model** | **Reason** |
|---------------|----------------|-------------|
| Real-time mobile sentiment | **DistilBERT** | Small & fast (~60% size of BERT). |
| Legal document research | **RoBERTa** | High accuracy via better pre-training. |
| Multilingual support | **XLM-RoBERTa** | Handles 100 + languages. |
| Efficient pretraining + token replacement | **ELECTRA** | Detect-replaced-token objective → efficient. |
| Resource-constrained environments | **ALBERT** | Parameter sharing → tiny memory footprint. |

---

## 2️⃣ Comparison Table

| **Model** | **Training Changes** | **Efficiency** | **Innovation** | **Use Cases** |
|------------|----------------------|----------------|----------------|----------------|
| **RoBERTa** | No NSP, dynamic masking, more data | Heavy | Longer training = better accuracy | Legal / finance text |
| **ALBERT** | Shared parameters + factorized embeddings | Very high | Smaller yet accurate | On-device / edge |
| **DistilBERT** | Distilled from BERT | High speed (~2× faster) | Knowledge distillation | Mobile / real-time |
| **ELECTRA** | Replaced-token detection | Very high | Generator + Discriminator training | Small data efficient |
| **XLM-RoBERTa** | Multilingual training on 100 langs | Large | Cross-lingual alignment | Global apps |

# 🌟 Exercise 6: Softmax Temperature — The Randomness Regulator

## 1️⃣ Temperature Scenarios

| **Temperature** | **Behavior** | **Description** |
|------------------|--------------|-----------------|
| **0.2** | Focused / Deterministic | Chooses safe high-probability tokens. |
| **1.0** | Balanced | Natural conversation flow. |
| **1.5** | Creative / Unpredictable | More diverse, risk of nonsense. |

---

## 2️⃣ Application Design

- **Bedtime Story Generator**  
  - Use **T ≈ 1.2–1.5** for imagination and fun.  
  - Lower to **0.8–1.0** to keep coherence and safety.  

- **Financial Report Summarizer**  
  - Use **T ≈ 0.2–0.5** for accuracy and precision.  
  - Avoid high T → prevents hallucinated numbers.  

---

## 3️⃣ Temperature and Bias

- **High T** → More diversity but may surface rare/biased phrases.  
- **Low T** → Repeats dominant language patterns → systemic bias.  

**Example:**  
Generating job ads:  
- At **1.4**, model may use biased descriptors (“strong male leader”).  
- At **0.3**, it may stick to corporate buzzwords that exclude diversity.  

**Mitigation:** bias filters, prompt controls, human review.
