# 🌟 Exercise 1 — Comparative Analysis of Generative AI vs Traditional AI

Critically evaluate which approach (Generative or Traditional AI) is most suitable in each scenario.  
Consider accuracy, creativity, efficiency, ethics, and cost.

---

### 1️⃣ Automated Medical Diagnosis
**Best fit:** Traditional AI (Classification model)  
**Reasoning:**
- High accuracy, interpretability, explainable Grad-CAM heatmaps.
- Generative models can **augment** training data (rare case synthesis) but not for final prediction.  
**Risks (Gen):** fake artifacts, liability issues.  
**Cost:** traditional cheaper at inference.

---

### 2️⃣ Legal Document Generation
**Best fit:** Hybrid → Generative LLM + rule-based validation.  
**Reasoning:**
- LLM drafts complex language; rules enforce legality and formatting.
- Add human-in-loop review for safety.  
**Risks:** hallucinated clauses, IP/data leakage.  
**Mitigations:** clause templates, retrieval-augmented generation (RAG), legal audits.

---

### 3️⃣ AI-Generated Scientific Research
**Best fit:** Generative AI (for summarization + idea generation).  
**Reasoning:**
- Excels at text compression, pattern mining.
- Must pair with citation checking and factual validation.  
**Risks:** fake citations, factual errors.  
**Mitigations:** citation databases, fact-checking, human peer review.

---

### 4️⃣ Financial Market Predictions
**Best fit:** Traditional AI (Time-series/Regression).  
**Reasoning:**
- Needs precision, explainability, backtesting.
- Generative AI useful only for stress-testing or scenario simulation.  
**Risks:** overfitting synthetic patterns.

---

### 5️⃣ Autonomous Vehicle Decision-Making
**Best fit:** Traditional AI (Perception + Planning Stack).  
**Reasoning:**
- Real-time reliability > creativity.
- Generative AI used only for simulation/edge-case generation.  
**Risks:** unpredictable generative behavior during driving.

---

# 🌟 Exercise 2 — Ethical & Security Risks of Generative AI

For each case: list ≥3 risks + 2 solutions.

---

### 1️⃣ Deepfake Political Manipulation
**Risks:** misinformation, voter deception, reputational harm.  
**Solutions:**  
- Digital watermarking + deepfake detection.  
- Election transparency laws + rapid removal protocols.

---

### 2️⃣ Synthetic Identity Fraud
**Risks:** fake KYC, financial theft, legal identity abuse.  
**Solutions:**  
- Liveness checks, multi-factor KYC, behavioral biometrics.  
- Dataset linking to verified gov records.

---

### 3️⃣ Generative AI in Cyber Warfare
**Risks:** false intel, policy errors, escalation.  
**Solutions:**  
- Provenance chains for intel sources.  
- Multi-agency fact verification.

---

### 4️⃣ AI-Generated Malware
**Risks:** undetectable mutations, rapid weaponization.  
**Solutions:**  
- Restrict model access + monitor usage.  
- Defensive Gen-AI for anomaly signature detection.

---

### 5️⃣ Copyright & IP Theft
**Risks:** plagiarism, artist income loss, lawsuits.  
**Solutions:**  
- Opt-in/opt-out datasets + royalty registries.  
- Model transparency + style protection filters.

# 🌟 Exercise 3 — Optimization & Fine-Tuning of Generative AI

---

### 1️⃣ Prompt Engineering (rewrite for specificity)
- “Generate an image of a futuristic city.”  
  → “4K concept art of a neon cyberpunk coastal city at night with maglev rails and holographic billboards.”
- “Write a poem about the future.”  
  → “12-line free-verse poem from a young engineer’s view of a climate-tech future.”
- “Create a song in the style of classical music.”  
  → “16-bar string quartet melody in A minor, tempo 92 BPM, ABA′ form.”

---

### 2️⃣ Bias & Fairness (dataset curation)
- Mix balanced sources (political spectrum, regions).  
- Debiasing via reweighting or counterfactual data.  
- Use fairness metrics and document data cards.

---

### 3️⃣ Fine-Tuning for Medical Research Papers
**Steps:**  
1. Curate peer-reviewed corpora (licensed).  
2. Clean + normalize citations.  
3. Apply transfer learning from strong LLM.  
4. Add RAG from PubMed.  
5. Evaluate: FactScore, BLEU, expert review.

---

### 4️⃣ Objective Evaluation Metrics
- **Text:** BLEU, ROUGE, BERTScore, Perplexity.  
- **Images:** FID, IS, CLIPScore, LPIPS.  
- **Audio:** MOS, VGGish-FD.  

---

### 5️⃣ Controlling Creativity & Coherence
- **Temperature:** lower = focused, higher = creative.  
- **RLHF:** train toward desired style or accuracy.  
- **Attention control:** anchor key phrases, prevent topic drift.

# 🌟 Exercise 4 — GANs vs VAEs Trade-Offs

---

### 1️⃣ Synthetic Medical Images

**Why:** Sharper realism for MRI; validate privacy.

---

### 2️⃣ AI-Assisted Creative Writing
**Pick:** VAE-style language model (if no LLM).  
**Why:** Structured latent space → better story coherence.

---

**Why:** Learns “normal” distribution; reconstruction error flags outliers.

**Why:** High-fidelity textures; use VAE for style interpolation.

# 🌟 Exercise 5 — Advanced Latent Space Exploration in VAEs

---

### 1️⃣ Visualizing Latent Space
- Encode test set → latent μ.  
- Visualize via t-SNE/UMAP.  
- Evaluate with Silhouette or NMI scores.

---

### 2️⃣ Interpolating Between Samples
- Encode digit 3 → z₃, digit 8 → z₈.  
- z(t) = (1−t)z₃ + t z₈ → decode each → smooth morph.  
- **Why VAE:** continuous latent + encoder; GANs lack inverse mapping.

---

### 3️⃣ Controlling Variability (KL Term)
- ↑ KL → smoother latent, less detail.  
- ↓ KL → sharper recon, risk overfit.  
- **Use case:** customized product variations.

---

### 4️⃣ Disentangling Latent Features
- Train β-VAE.  
- Identify latent zᵢ for specific attributes (e.g., hair color).  
- Modify only that dimension → isolated change.

---

### 5️⃣ PCA vs VAE Comparison

| Aspect | PCA | VAE |
|:--|:--|:--|
| Linearity | Linear | Non-linear |
| Components | Orthogonal axes | Neural latent dims |
| Reconstruction | Sharp for linear data | Handles complex manifolds |
| Generative | ❌ | ✅ (sample z→decode) |
