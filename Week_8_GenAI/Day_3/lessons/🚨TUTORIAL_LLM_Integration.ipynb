```python
import requests
import os

# Set your Google Cloud API key in the environment
# Corrected: Use the second argument of os.getenv() for a default value,
# or set the environment variable first.
# It's recommended to use Colab secrets for sensitive keys:
# from google.colab import userdata
# API_KEY = userdata.get('YOUR_API_KEY_SECRET_NAME')

# If not using Colab secrets, define your API key like this (less secure for sharing):
API_KEY = os.getenv("GCLOUD_API_KEY", "GEMINI_API_KEY") # Assigns the env var or the default value

# Ensure the PROJECT_ID in the API_URL is correct for your Google Cloud Project.
# The original URL had a hardcoded project ID that might not be yours.
# Replace 'gen-lang-client-0880816886' with your actual Google Cloud Project ID.
PROJECT_ID = "gen-lang-client-0880816886" # <-- Replace with your actual Project ID if different

API_URL = (
    f"https://us-central1-aiplatform.googleapis.com/v1/projects/"
    f"{PROJECT_ID}/locations/us-central1/publishers/google/models/gemini-pro:generateText"
)

headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {API_KEY}"
}

def generate_tweet(blog_text: str) -> str:
    payload = {
        "instances": [
            {
                "prompt": f"Create a tweet from this blog post:\n{blog_text}"
            }
        ],
        "parameters": {
            "temperature": 0.7,
            "maxOutputTokens": 100
        }
    }
    resp = requests.post(API_URL, headers=headers, json=payload)
    resp.raise_for_status()
    return resp.json()["predictions"][0]["content"]
```
