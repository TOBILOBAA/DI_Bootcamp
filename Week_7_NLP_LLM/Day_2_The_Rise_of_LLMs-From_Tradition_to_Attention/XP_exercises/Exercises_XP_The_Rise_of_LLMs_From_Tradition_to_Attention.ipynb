# **Exercise 1 : What are Large Language Models (LLMs)?**

LLMs are machine learning models trained on series of text and data to understand certain patterns that can be used to make predictions. They are designed to be the brain of AI to improve accuracy and efficiency.


```python
# Install necessary libraries
!pip install transformers matplotlib --quiet

# Import required libraries
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
# 2. Loading a pretrained model and tokenizer
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, num_labels=3)

print(f"\nModel '{model_name}' loaded successfully!")
print("""
GPT-2 is a causal language model, meaning it predicts the next word in a sequence.
It has been trained on a diverse dataset and can generate coherent, contextually relevant text.
""")
```

# **Exercise 2 : Transformer Architecture and Tokenization**

Tokenization means splitting text into pieces and turning those pieces into numbers so the model can process and learn from them.


```python
# 2. Tokenize input text
text = "This is a sample sentence for tokenization"
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# 3. Visualizing the tokenization process
plt.figure(figsize=(10, 4))
plt.bar(tokens, token_ids, color="skyblue")
plt.xlabel("Tokens")
plt.ylabel("Token IDs")
plt.title("Visualization of Tokenization Process")
plt.show()

```

# **Exercise 3 : Understanding Token IDs and Special Prefixes**

```python
print(f"Original Text: {text}")
print(f"Tokens: {tokens}")
print(f"Token IDs: {token_ids}")
```

# **Exercise 4 : Pretraining vs. Fine-Tuning**

Pretraining is teaching the computer from text, journal articles, and other resources to understand patterns. Fine-tuning is taking an already trained model and tweaking it to fit specific needs.”

# **Exercise 5 : Generate Simple Text**

```python
input_text = "Well this is another sample sentence i'm creating how about that."
input_ids = tokenizer.encode(input_text, return_tensors="pt")
output_ids = model.generate(input_ids, max_length=50, do_sample=True)
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

print(f"Input: {input_text}")
print(f"Generated Output: {output_text}")
```
