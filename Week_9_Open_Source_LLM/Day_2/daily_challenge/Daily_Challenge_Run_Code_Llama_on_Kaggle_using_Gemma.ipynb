```python
import os
os.makedirs("/content/models", exist_ok=True)

# TinyLlama 1.1B Chat
!wget -q https://huggingface.co/TheBloke/TinyLlama-1.1B-Chat-v1.0-GGUF/resolve/main/tinyllama-1.1b-chat-v1.0.Q5_K_M.gguf \
     -O /content/models/tinyllama-1.1b-chat-v1.0.Q5_K_M.gguf
MODEL_PATH = "/content/models/tinyllama-1.1b-chat-v1.0.Q5_K_M.gguf"
```

# **Load the model and generate example**

```python
from llama_cpp import Llama
import os

# Heuristics: push some layers to GPU if available (safe defaults)
# has_gpu = bool(subprocess.getoutput("nvidia-smi -L"))
n_gpu_layers = 40   # if has_gpu else 0   # tweak up/down based on VRAM
n_threads = max(1, os.cpu_count() // 2)

llm = Llama(
    model_path=MODEL_PATH,
    n_gpu_layers=n_gpu_layers,
    n_threads=n_threads,
    logits_all=False,
    verbose=False
)

# ---- Text generation example ----
prompt = "Explain how the solar system formed in 3 short bullet points."
output = llm(
    prompt=prompt,
    max_tokens=256,
    temperature=0.7,
    top_p=0.95,
)
print(output["choices"][0]["text"])

# ---- Streaming code generation example ----
stream = llm(
    prompt=(
        "Write a Python script that loads a Hugging Face model and tokenizes input. "
        "Keep it under 25 lines."
    ),
    max_tokens=300,
    temperature=0.7,
    top_p=0.95,
    stream=True,
)

for chunk in stream:
    token = chunk["choices"][0]["text"]
    print(token, end="", flush=True)
```
