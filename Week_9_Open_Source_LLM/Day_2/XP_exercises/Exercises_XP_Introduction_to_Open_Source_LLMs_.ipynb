# **Exercise 1: Open Source Levels Reflection**

## Step 1: Gather Definitions

### 🔹 Fully Open
The entire model—including code, architecture, and pretrained weights—is publicly available.  
Users can inspect, modify, retrain, or repurpose it without restriction (depending on license).

### 🔹 Weights Released
Only the pretrained weights are provided.  
The model architecture might be known, but the full training pipeline or dataset is not open.

### 🔹 Architecture Only
Only the model’s structure (layers, parameters, design) is published.  
No pretrained weights are provided, so you must train from scratch.

---

## Step 2: Identify Key Characteristics

| Openness Level | What’s Open? | What You Can / Can’t Do |
|----------------|--------------|--------------------------|
| **Fully Open** | Source code, architecture, and pretrained weights | ✅ Retrain, fine-tune, inspect internals. ❌ Must still check license for commercial use. |
| **Weights Released** | Pretrained weights (no full training data or scripts) | ✅ Fine-tune for new tasks. ❌ Cannot fully reproduce training. |
| **Architecture Only** | Network design only | ✅ Study structure or re-implement. ❌ No pretrained performance; must train from scratch. |

---

## Step 3: Comparative Paragraph

Fully open models provide total transparency, allowing developers to inspect, modify, and retrain both the code and pretrained weights—ideal for research and innovation.  
Weights-released models share the pretrained parameters but keep training data and pipeline private, which still enables fine-tuning but limits reproducibility.  
Architecture-only models share just the blueprints, offering learning value but requiring costly retraining.  
Each level trades openness for control, privacy, and intellectual-property protection.

---

## Step 4: Healthcare Prompt

For building a **healthcare-specific assistant**, the **Weights Released** or **Fully Open** level is essential, since clinical data fine-tuning requires access to pretrained weights.  
This level allows domain adaptation while maintaining compliance and avoiding retraining from scratch on sensitive data.

```bash
brew install cmake make git

# **Exercise 5: Benchmark-Based Model Explorer**

## Step 1: Open the Leaderboard
Visit the **[Open LLM Leaderboard](https://huggingface.co/spaces/HuggingFaceH4/open_llm_leaderboard)**.  
This leaderboard tracks public open-source LLMs using two key benchmarks:
- **HellaSwag:** Measures commonsense reasoning ability.  
- **MMLU:** Evaluates multi-task academic understanding across 57 subjects.

---

## Step 2: Select Three Models
We’ll pick three that represent different strengths:  
- One strong in commonsense reasoning.  
- One strong in academic understanding.  
- One balanced performer.

| Purpose | Model | Link |
|----------|--------|------|
| Commonsense reasoning | `mistralai/Mistral-7B-Instruct` | [View model](https://huggingface.co/mistralai/Mistral-7B-Instruct) |
| Academic / multi-task | `meta-llama/Llama-3-8B-Instruct` | [View model](https://huggingface.co/meta-llama/Llama-3-8B-Instruct) |
| Balanced multilingual | `bigscience/bloomz-7b1` | [View model](https://huggingface.co/bigscience/bloomz-7b1) |

---

## Step 3: Record Benchmark Scores

| Model | HellaSwag Score | MMLU Score | License |
|--------|-----------------|-------------|----------|
| `mistralai/Mistral-7B-Instruct` | ~85.0 | ~63.0 | Apache 2.0 |
| `meta-llama/Llama-3-8B-Instruct` | ~88.7 | ~74.5 | Custom Llama 3 Community License |
| `bigscience/bloomz-7b1` | ~72.0 | ~57.0 | RAIL License 1.0 |

*(Scores approximate from the latest leaderboard snapshot — use current values when reporting.)*

---

## Step 4: Analyze License Implications

| License Type | Description | Commercial Use |
|---------------|--------------|----------------|
| **Apache 2.0** | Fully permissive; allows modification, redistribution, and commercial use. | ✅ Allowed |
| **Llama 3 Community License** | Requires model registration and responsible use; commercial use permitted with compliance. | ✅ Conditional |
| **RAIL 1.0** | Restrictive license focused on ethical use; limits certain high-risk applications. | ⚠️ Restricted |

---

## Step 5: Ideal Use-Case Mapping

| Model | Ideal Use Case | Reason |
|--------|----------------|--------|
| `mistralai/Mistral-7B-Instruct` | **Customer support / chatbots** | Fast inference, strong commonsense reasoning, permissive license. |
| `meta-llama/Llama-3-8B-Instruct` | **Academic tutor / enterprise assistant** | High MMLU score, scalable accuracy across tasks. |
| `bigscience/bloomz-7b1` | **Multilingual NGO chatbot** | Fluent in 40+ languages, open collaboration model. |

---

## Step 6: Comparison Table Summary

| Model | HellaSwag | MMLU | License | Ideal Use Case |
|--------|-----------|------|----------|----------------|
| `mistralai/Mistral-7B-Instruct` | 85.0 | 63.0 | Apache 2.0 | Commonsense Q&A agent |
| `meta-llama/Llama-3-8B-Instruct` | 88.7 | 74.5 | Custom (Llama 3 Community License) | Academic reasoning tutor |
| `bigscience/bloomz-7b1` | 72.0 | 57.0 | RAIL 1.0 | Multilingual global assistant |

---

### 🧠 Reflection
Benchmarks provide objective insight beyond marketing claims.  
- **HellaSwag** helps identify models strong in real-world reasoning.  
- **MMLU** reflects a model’s academic and logical depth.  
Selecting models based on benchmark data ensures your choice aligns with use-case needs rather than hype — improving reliability and ROI.

---

**Deliverables**
1. ✅ Model list and links.  
2. ✅ Completed comparison table with benchmark and license info.  
3. ✅ Reflection paragraph on benchmark importance.

# **Exercise 6: Cloud vs Local Deployment Plan**

## Step 1: Review Comparison
Before choosing where to deploy (cloud vs local), it’s important to weigh **cost, control, performance, scalability, and security**.  
Below is a simple breakdown of the trade-offs between running models **locally on your machine** and **in the cloud** (like Google Colab, RunPod, or AWS).

---

## Step 2: Deployment Comparison

| Category | Local Deployment | Cloud Deployment |
|-----------|------------------|------------------|
| **Cost** | ✅ One-time hardware investment (no recurring fees).<br>❌ Expensive upfront and limited upgrades. | ✅ Pay-as-you-go — scalable GPU access.<br>❌ Long-term cost can grow quickly. |
| **Performance & Latency** | ✅ Faster response (no internet delay).<br>❌ Limited by your CPU/GPU specs. | ✅ Access to high-end GPUs (A100, T4, etc.).<br>❌ Slower response if internet is weak. |
| **Control & Privacy** | ✅ Full control of data and model.<br>✅ No external dependency. | ❌ Data stored remotely.<br>✅ Easier to share and collaborate securely. |
| **Scalability** | ❌ Hard to scale beyond one machine. | ✅ Easily scale to multiple instances. |
| **Ease of Setup** | ❌ Requires technical setup (environment, dependencies). | ✅ Pre-configured notebooks (Colab, RunPod). |

---

## Step 3: 5-Bullet Pros & Cons Summary

- ✅ **Local advantage:** Full data privacy and zero ongoing costs.  
- ⚠️ **Local downside:** Needs strong hardware (≥16GB RAM, GPU for 7B+ models).  
- ✅ **Cloud advantage:** Ready access to high-end GPUs and scalable infrastructure.  
- ⚠️ **Cloud downside:** Pay-per-hour cost and data privacy trade-offs.  
- 💡 **Hybrid approach:** Develop locally, deploy or fine-tune in the cloud.

---
