🚑 AI Agent for Emergency Medical Dispatch — Design Document

👤 1. Understanding the Scenario

The goal is to design an AI-powered agent that supports emergency medical dispatchers by:
	•	Triaging incoming 911 calls
	•	Understanding symptoms and severity
	•	Recommending actions or dispatching emergency services
	•	Maintaining accurate state throughout the call
	•	Using external tools (APIs/models) to make safe decisions

⸻

🌍 2. Environment Definition (Agent Perception Layer)

The agent receives the following inputs:

📥 Caller Inputs
	•	Text transcript of the caller’s symptoms (from speech-to-text or direct text)
	•	Tone/keywords indicating distress or danger
	•	Follow-up answers to the agent’s clarifying questions

📍 Location Data
	•	GPS coordinates
	•	Street address
	•	Distance to nearest hospital/ambulance

📚 Medical Context
	•	Caller identity (if available)
	•	Known medical history (e.g., allergies, chronic disease)
	•	Past emergency call records

⸻

🛠 3. Tools & External APIs

The agent integrates with several external systems:

1. Symptom Checker API (e.g., Infermedica)
	•	Input: List of symptoms, durations, severity keywords
	•	Output: Possible conditions + severity score + red-flag indicators

2. Ambulance Dispatch / Scheduling System
	•	Input: Severity score, caller location, required medical team
	•	Output: Confirmation of dispatch, estimated arrival time

3. GIS / Location Services
	•	Input: Caller GPS, type of emergency
	•	Output: Nearest hospital, optimal route, response times

4. Medical Triage LLM (fine-tuned model)
	•	Input: Symptom summary, medical history
	•	Output: Numerical urgency score (0–1) + recommended action

⸻

🧠 4. State Management (Memory System)

The agent maintains persistent state throughout the call:

{
  "caller_id": null,
  "contact_number": null,
  "location": {
    "address": "",
    "gps_coordinates": ""
  },
  "symptoms": [],
  "severity_keywords": [],
  "urgency_score": 0,
  "triage_level": "",
  "action_taken": "",
  "decision_log": []
}

The agent MUST remember:
	•	Identity + location of caller
	•	All symptoms reported
	•	Severity score from API/LLM
	•	Decisions taken so far
	•	Whether an ambulance has already been dispatched

If this fails, the agent could repeat questions, miss critical signs, or send incorrect services.

⸻

🧩 5. Decision-Making Workflow (Reasoning Loop)

Step-by-step triage flow:

1. Receive transcript from 911 caller.
2. Extract symptoms + severity words (e.g., "can’t breathe", "chest pain").
3. Query Symptom Checker API → returns severity indicators.
4. Query Triage LLM → compute urgency score (0–1).
5. Compare urgency score to thresholds:
     - High (≥ 0.75) → IMMEDIATE AMBULANCE
     - Medium (0.40–0.74) → Recommend urgent care
     - Low (< 0.40) → Provide self-care guidance
6. Use dispatch system if needed.
7. Update state log with final decision.

Flowchart (text version)

Caller Input → Symptom Extraction → Severity Model → Urgency Score
        ↓                                    ↓
Update State  ←———— Decision Thresholds ———→ Action:
                                          - Dispatch ambulance
                                          - Urgent care advice
                                          - Self-care steps


⸻

🧠 6. Agent Classification

Chosen Architecture: Hybrid Agent

A Hybrid Agent combines:
	•	Reactive behavior: immediate response to life-threatening symptoms
	•	Deliberative reasoning: planning + consulting APIs/models before deciding

Justification
	•	Emergency triage requires both fast reaction (e.g., “not breathing”) and careful planning (evaluating all symptoms and history).
	•	The agent must use memory/state to avoid repeating questions.
	•	It integrates multiple tools, which requires deliberate decision sequences.

⸻

🔄 7. Comparison to Another Agent Type: Reactive Agent

Reactive Agent Characteristics

A reactive agent:
	•	Only responds to the immediate input
	•	Doesn’t plan or use multi-step reasoning
	•	Doesn’t rely on memory
	•	Perfect for simple tasks — not medical dispatch

Differences if this system were Reactive:

Topic	Hybrid Agent	Reactive Agent
Memory	Remembers caller info	Forgets everything each step
Planning	Uses multi-step analysis	Acts instantly on latest message
Tools	Chooses tools based on reasoning	May call tools incorrectly or too often
Decisions	Consistent, safe	Risky, impulsive

Trade-offs
	•	Speed: Reactive faster ⚡
	•	Reliability: Hybrid more accurate and safe
	•	Intelligence: Hybrid provides nuanced medical judgment

⸻

🪬 8. Reflection

What fails without state?

Without state:
	•	The agent might forget symptoms
	•	Could ask the caller the same questions repeatedly
	•	Might dispatch the wrong emergency service
	•	Decisions would contradict earlier steps
This is unsafe in medical emergencies.

Why are tools essential?
	•	No LLM alone can determine medical urgency reliably
	•	Tools provide real medical evidence, real-time data, and validated scoring
	•	GIS tools ensure the fastest dispatch route
	•	Symptom APIs catch red-flag situations an LLM may miss

Tools convert the agent from “smart text generator” → safe, medical-grade system.
