🌟 Exercise 1: Rewrite And Optimize A Vague Prompt

1.1 Three Critical Issues with the Prompt

Original prompt:

“Write something about productivity tips.”

Issues:
	1.	No target audience
	•	We don’t know who this is for (students, remote workers, managers, etc.).
	2.	No brand or product context
	•	It doesn’t mention FlowNest, the focus app we’re actually promoting.
	3.	No platform / format guidance
	•	It doesn’t say it’s a LinkedIn post, what tone to use, or the structure/length (hook, CTA, etc.).

⸻

1.2 Rewritten Prompt (Using Role + Instructions + Constraints)

Act as a senior SaaS marketing copywriter specializing in productivity tools.

Write a LinkedIn post promoting our new focus app **FlowNest** to busy knowledge workers (remote employees, freelancers, and startup teams).

- Give **3 actionable productivity tips** that FlowNest can help with (e.g., deep work sessions, reduced distractions).
- Connect each tip to **one feature** of FlowNest.
- End with a clear **call-to-action** to try FlowNest.

Constraints:
- Tone: friendly, practical, and slightly aspirational.
- Format: short intro hook, then **3 bullet points**, then a 1–2 sentence CTA.
- Length: maximum **180–220 words**.

Use this structure and tone whenever we promote productivity tools on social media for FlowNest.

🌟 Exercise 2: Multi-Part Prompt For Quiz Generation

2.1 Complete Prompt for Quiz Generation

You are an experienced 7th-grade science teacher.

I will give you a short article about **volcanic eruptions**.  
Using that article only, do the following:

1. **Summary**
   - Summarize the article in **2 simple bullet points**.
   - Use clear, friendly language that 11–13-year-olds can understand.

2. **Quiz Questions**
   - Create **3 multiple-choice questions** about the key ideas.
   - For each question:
     - Write **1 correct answer** and **2 plausible distractors**.
     - Label the correct answer clearly (e.g., “Correct answer: B”).

3. **Tone & Vocabulary**
   - Use a **friendly, encouraging tone**.
   - Keep vocabulary **simple and age-appropriate** for 7th graders.

4. **Output Format (for Google Slides)**
   - Structure the final output as Google Slides content:
     - **Slide 1:** Title + 2 summary bullet points.
     - **Slide 2:** Question 1 + answer options + indicate the correct answer.
     - **Slide 3:** Question 2 + answer options + correct answer.
     - **Slide 4:** Question 3 + answer options + correct answer.

Do not add extra facts beyond what is in the article.
Ask me to paste the article when you are ready.


⸻

2.2 Comparison: Why This Is Better Than “Make a quiz for kids about this article.”

Vague prompt:

“Make a quiz for kids about this article.”

Three key improvements:
	1.	Clear structure & tasks
	•	My prompt explicitly asks for summary + questions, and a slide structure, so the output is organized and ready for Google Slides.
	2.	Age-appropriate control
	•	It specifies 7th-grade, simple vocabulary, and a friendly tone, reducing the chance of content being too hard or too technical.
	3.	Precise format for MCQs
	•	It defines number of questions, answer structure, and labeling correct answers, so the quiz is immediately usable without manual fixing.

🌟 Exercise 3: Add Context, Get Better Results

Original intern prompt:

“Summarize this report.”

3.2 Rewritten Prompt with Context

Act as a senior financial analyst.

You are helping me prepare a **3-minute spoken summary** of our **monthly finance report** for a **non-technical executive leadership team**.

Using only the content of this report:

- Extract **3 key financial takeaways** (e.g., revenue, major cost changes, profit, key variances).
- For each takeaway:
  - Include at least **one concrete number** (percentage, amount, or trend).
  - Briefly explain what it means in **plain, non-technical language**.

Format:
- Output as **3 bullet points**.
- Under **180–200 words** total.
- Write it as if it were a **speech script** I can read out loud in a meeting.

Do not invent any numbers; use only data found in the report.
Ask me to paste the report when you are ready.

🌟 Exercise 4: Match Prompt To Purpose

Let’s choose the Structured style (best for consistent customer support responses).

4.1 Structured Prompt for an E-Commerce Support Chatbot

You are a customer support assistant for an e-commerce store that sells clothing and accessories.

Your goals:
- Quickly understand the customer’s issue.
- Provide a clear, polite, and practical answer.
- Ask follow-up questions only when necessary.

When the user sends a message, respond in this structure:

1. **Brief Acknowledgment** (1 sentence)
   - Empathize or confirm you understand the issue.

2. **Main Answer / Solution**
   - Explain the answer in **2–4 short bullet points**.
   - If it’s about:
     - **Order status** → Explain how to track, expected timelines.
     - **Returns/refunds** → Summarize the policy and next steps.
     - **Sizing/product info** → Provide guidance and link to size charts or product details if known.

3. **Next Step / CTA**
   - End with **one clear next step** (e.g., “Click here…”, “Reply with your order number…”).

Constraints:
- Be polite and professional.
- Avoid jargon.
- If you don’t have enough info, **ask 1–2 specific follow-up questions** instead of guessing.


⸻

4.2 Why This Matches the Structured Style

Key features:
	1.	Explicit sections: Acknowledgment → Answer → Next Step.
	2.	Fixed format with bullets: Ensures responses are predictable, scannable, and on-brand.

⸻

4.3 Why Structured Is Better Here than Other Styles
	•	Better than Exploratory:
Customer support needs clarity and consistency, not brainstorming or idea generation.
	•	Better than Conversational-only:
Chatty replies without structure can feel slow and messy for customers needing quick help.
	•	Better than pure Functional:
Functional is great for tasks like “summarize this,” but support needs a mix of empathy + clarity + action in a repeatable template.

🌟 Exercise 6: Hallucination Spotting And Mitigation

6.1 Precise Prompt to Avoid Hallucinations (Version 1)

You are summarizing a peer-reviewed article on climate change and marine biodiversity.

Rules:
- Use **only** information that appears in the provided article.
- **Do not invent** numbers, statistics, or claims.
- If a detail is not clearly stated in the text, **do not include it**.

Task:
- Write a concise summary (150–200 words) of the article’s main findings about climate change impacts on marine biodiversity.
- Focus on: observed effects, studied regions, and any mitigation or adaptation strategies mentioned.

If you are unsure about a fact, **leave it out** rather than guessing.


⸻

6.2 Prompt with Explicit Verification Language (Version 2)

You are summarizing a peer-reviewed article on climate change and marine biodiversity.

Verification rules (very important):
- Only include claims, numbers, and conclusions that are **explicitly stated** in the article.
- **Do not infer or guess** extinction rates, future projections, or numerical impacts.
- If you cannot confirm a detail from the text, write:  
  **“This information is not specified in the article.”**

Tasks:
1. Write a 150–200 word summary of the article’s main points.
2. List **3 key findings** in bullet points.
3. For each bullet, include a short quote or phrase from the article that supports it (no more than 10 words).

Do not add external knowledge or assumptions beyond the article.


⸻

6.3 Mitigation Strategies Used

Two strategies in the prompts above:
	1.	Source-bounded instruction
	•	Explicitly telling the model to use only the provided article and avoid external knowledge.
	2.	Uncertainty handling
	•	Instructing the model to say “This information is not specified in the article.” when unsure, instead of guessing.

Other things you could add (if needed): ask for inline citations, or require it to quote the exact sentence for each claim.

⸻

6.4 Domains Where Hallucinations Are Especially Risky
	1.	Healthcare / Medicine
	•	Hallucinated treatment recommendations, dosages, or diagnoses can directly harm patients, delay real care, or cause dangerous self-medication.
	2.	Legal / Compliance
	•	Fabricated laws, cases, or regulations can lead to faulty contracts, non-compliance with regulations, fines, lawsuits, or serious legal exposure for companies and individuals.
