⭐ Step 5b — Questions to Answer

When is quoting more appropriate than paraphrasing?
	•	When legal, compliance, or policy wording must remain exact
	•	When clarity requires using the organization’s official approved phrasing
	•	When the communication must avoid misinterpretation or accidental rewriting
	•	When emphasizing the seriousness or authority of a rule

When might quoting pose a risk?
	•	When the original text is too technical or formal for general employees
	•	When the language is full of jargon that reduces understanding
	•	When the exact wording could be misinterpreted without context
	•	When a simplified, employee-friendly paraphrase is safer and clearer
