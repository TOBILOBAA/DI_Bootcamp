**Instructions**

Which model(s) would you pick for a Loan Prediction ?

Outline the steps to evaluate the model’s performance, mentioning specific

metrics that would be relevant to evaluate the model.

I would use an unsupervised learning model, specifically K-Means clustering. Since the dataset does not directly provide a label indicating whether an applicant defaulted or not, clustering helps group applicants into segments (e.g., high-risk vs low-risk borrowers) based on similarities in their financial and demographic features. This allows us to detect hidden patterns that may signal default risk.

**How would I evaluate the model’s performance?**

For an unsupervised model like K-Means, I would first split the dataset into a training set (to fit the clusters) and a testing set (to check consistency of the cluster patterns). To evaluate performance, I would use metrics that measure the quality of clusters, such as:

	•	Silhouette Score → checks how well applicants fit within their cluster compared to other clusters.

	•	Davies–Bouldin Index → measures how separated and distinct the clusters are (lower values are better).
  
	•	Visual inspection → using dimensionality reduction (e.g., PCA) to plot the clusters and see if they are interpretable and meaningful.

**Exercise 5: Designing an Evaluation Strategy**

Supervised Model (Classification – e.g., Decision Tree / Logistic Regression)

	•	Strategy: Split the dataset into training and testing sets (e.g., 80/20). Use k-fold cross-validation to reduce bias.
  
	•	Metrics:

      •	Accuracy → overall correctness.

      •	Precision & Recall → important if misclassification costs differ (e.g., false positives vs false negatives).

      •	F1-Score → balances precision and recall.

      •	ROC Curve & AUC → measures how well the model distinguishes between classes.


Unsupervised Model (Clustering – e.g., K-Means)

	•	Strategy: No labels exist, so evaluation looks at cluster quality.
  
	•	Metrics:
      •	Silhouette Score → measures how well data points fit their own cluster vs. others (higher is better).

      •	Davies–Bouldin Index → checks separation of clusters (lower is better).

      •	Elbow Method → helps decide the optimal number of clusters.

      •	Visual Inspection: Use PCA/t-SNE to plot clusters and check if they make sense.


Reinforcement Learning Model (e.g., Q-Learning for navigation)

	•	Strategy: Train the agent in a simulated environment and track its performance over episodes.

	•	Metrics:

      •	Cumulative Reward → total rewards earned over time.

      •	Convergence → how quickly the policy stabilizes.

      •	Exploration vs Exploitation → balance between trying new actions and using learned strategies.


4. Challenges & Limitations

	•	Supervised → risk of overfitting, biased datasets, metrics may give conflicting signals.

	•	Unsupervised → no “ground truth” labels, so evaluation is more subjective.

	•	Reinforcement → training can be slow, environment design and reward setting strongly influence results.
